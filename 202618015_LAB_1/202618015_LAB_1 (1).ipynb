```python
import pandas as pd
import numpy as np
```

```python
d = pd.read_csv("books.csv")
d.head()
```

```python
# checking for missing value
print("Total Records:", len(d))

print("\nMissing Values:")
print(d.isnull().sum())

print("\nDuplicate UPCs:")
print(d["UPC"].duplicated().sum())
```

```python
# removing extra spaces 
d["title"] = d["title"].str.strip()
d["category"] = d["category"].str.strip()
d["availability"] = d["availability"].str.strip()
d["product_description"] = d["product_description"].str.strip()

d.head()
```

```python
d = d.drop_duplicates(subset="UPC")
print("Total Records after removing duplicates:", len(d))
```

```python
# converting price to numeric values
d["price"] = d["price"].str.replace("£","", regex=False)
d["price"] = d["price"].astype(float)
d["price"].head()
```

```python
rating_map = {

    "One":1,

    "Two":2,

    "Three":3,

    "Four":4,

    "Five":5

}

d["rating"] = d["rating"].map(rating_map)

d[["rating"]].head()
```

```python
d["stock_count"] = d["availability"].str.extract(r'(\d+)')
d["stock_count"] = d["stock_count"].fillna(0)
d["stock_count"] = d["stock_count"].astype(int)
d[["availability","stock_count"]].head()
```

```python
d["description_word_count"] = d["product_description"].apply(lambda x: len(str(x).split()))
d[["description_word_count"]].head()
```

```python
d["price_band"] = pd.cut(

    d["price"],

    bins=[0,20,40,100],

    labels=["Low","Medium","High"]

)

d[["price","price_band"]].head()
```

```python
d["affordability_score"] = round(d["rating"] / d["price"],2)

d[["price","rating","affordability_score"]].head()
```

```python
d["recommended"] = np.where(

    (d["rating"]>=4) & (d["price"]<35),

    "Yes",

    "No"

)

d[["title","recommended"]].head()
```

```python
d.to_csv("cleaned_books.csv",index=False)
print("Cleaned dataset saved successfully.")
```
