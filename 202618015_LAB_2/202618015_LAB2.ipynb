```python
# TASK 2
#1
import numpy as np

# Create two 2x2 matrices
A = np.array([
    [2, 3],
    [4, 5]
])

B = np.array([
    [6, 7],
    [8, 9]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
```

```python
#2
addition = A + B

print("A + B:")
print(addition)
```

```python
#3
elementwise_multiplication = A * B

print("A * B:")
print(elementwise_multiplication)
```

```python
#4
matrix_multiplication = A @ B

print("A @ B:")
print(matrix_multiplication)
```

```python
#5
transpose_A = np.transpose(A)

print("Transpose of A:")
print(transpose_A)
```

```python
#6
det_A = np.linalg.det(A)

print("Determinant of A:", det_A)
```

```python
#7
if not np.isclose(det_A, 0):
    inverse_A = np.linalg.inv(A)

    print("Inverse of A:")
    print(inverse_A)
else:
    print("Matrix A is not invertible.")
```

```python
#8
verification = A @ inverse_A

print("A @ A^-1:")
print(verification)
```

```python
is_identity = np.allclose(verification, np.eye(A.shape[0]))

print("Is A @ A^-1 an identity matrix?", is_identity)
```

```python
# TASK 3
#1
import numpy as np
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

# Chosen population parameters
mean = 50
std = 10

# Generate 1000 values from a normal distribution
data = np.random.normal(loc=mean, scale=std, size=1000)

print("Number of values:", len(data))
print("First 10 values:")
print(data[:10])
```

```python
#2
# Calculate sample mean
sample_mean = np.mean(data)

# Calculate sample standard deviation
sample_std = np.std(data, ddof=1) # DDOF TAKES VALUE 0 AND 1 FOR SAMPLE AND POPULATION STANDARD DEVIATION RESPECTIVELY

print("Chosen Mean:", mean)
print("Sample Mean:", sample_mean)

print("\nChosen Standard Deviation:", std)
print("Sample Standard Deviation:", sample_std)
```

```python
print("Mean Difference:",
      sample_mean - mean)

print("Standard Deviation Difference:",
      sample_std - std)
```

```python
#4
plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, edgecolor='black')
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normally Distributed Data")
plt.show()
```

```python
#4
df_original = pd.read_csv("train.csv")
```

```python
#A. MEAN IMPUTATION
df_mean = df_original.copy()

mean_age = df_mean['Age'].mean()

df_mean['Age'] = df_mean['Age'].fillna(mean_age)

print("Mean imputation:")
print("Missing Age:", df_mean['Age'].isnull().sum())
print("Value used:", mean_age)
```

```python
#B. MEDIAN IMPUTATION
df_median = df_original.copy()

median_age = df_median['Age'].median()

df_median['Age'] = df_median['Age'].fillna(median_age)

print("Median imputation:")
print("Missing Age:", df_median['Age'].isnull().sum())
print("Value used:", median_age)
```

```python
#C. MODE IMPUTATION
df_mode = df_original.copy()

mode_age = df_mode['Age'].mode()[0]

df_mode['Age'] = df_mode['Age'].fillna(mode_age)

print("Mode imputation:")
print("Missing Age:", df_mode['Age'].isnull().sum())
print("Value used:", mode_age)
```

```python
#D. RANDOM VALUE IMPUTATION
import numpy as np
df_random = df_original.copy()

np.random.seed(42)

missing_mask = df_random['Age'].isnull()

observed_age = df_random.loc[
    df_random['Age'].notnull(),
    'Age'
]

df_random.loc[missing_mask, 'Age'] = np.random.choice(
    observed_age,
    size=missing_mask.sum(),
    replace=True
)

print("Random-value imputation:")
print("Missing Age:", df_random['Age'].isnull().sum())
```

```python
#5
comparison = pd.DataFrame({
    'Method': ['Mean', 'Median', 'Mode', 'Random'],
    'Imputed Value': [
        mean_age,
        median_age,
        mode_age,
        'Random observed ages'
    ],
    'Missing After Imputation': [
        df_mean['Age'].isnull().sum(),
        df_median['Age'].isnull().sum(),
        df_mode['Age'].isnull().sum(),
        df_random['Age'].isnull().sum()
    ]
})

print(comparison)
```
