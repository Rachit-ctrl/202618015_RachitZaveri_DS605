## TASK 1

```python
import pandas as pd
import numpy as np
import pickle
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv('AB_NYC_2019.csv')
data.head()
```

```python
print("Missing values before cleaning:")
print(data.isnull().sum())
```

```python
data['name'] = data['name'].fillna('Unknown')
data['host_name'] = data['host_name'].fillna('Unknown')
```

```python
data['reviews_per_month'] = data['reviews_per_month'].fillna(0)
```

```python
columns_to_drop = ['id', 'host_id', 'last_review', 'name', 'host_name']
data = data.drop(columns=columns_to_drop)
```

```python
data = data[data['price'] > 0]
data = data[data['price'] < 1000]
```

```python
data = pd.get_dummies(data, columns=['neighbourhood_group', 'neighbourhood', 'room_type'], drop_first=True)
```

```python
Y = data['price']
X = data.drop(columns=['price'])

print(X.shape)
print(Y.shape)
```

## TASK 2

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
train_X, test_X, train_Y, test_Y = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
model1 = LinearRegression()
model1.fit(train_X, train_Y)
predictions1 = model1.predict(test_X)
print("Linear Regression Error (MAE):", mean_absolute_error(test_Y, predictions1))
```

```python
model2 = DecisionTreeRegressor(max_depth=10, random_state=42)
model2.fit(train_X, train_Y)
predictions2 = model2.predict(test_X)
print("Decision Tree Error (MAE):", mean_absolute_error(test_Y, predictions2))
```

```python
model_data = {
    'model': model2,
    'columns': list(train_X.columns)
}
```

```python
with open('airbnb_model.pkl', 'wb') as file:
    pickle.dump(model_data, file)
```
